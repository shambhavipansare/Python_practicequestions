# Nested List Indexing

# 1. Accessing Triple-Nested Items

```python
nested = [1, [2, 3, [4, 5, 6]], 7]
```

```python
# access 2

nested[1][0]
```

```python
# access [4, 5, 6]

nested[1][2]
```

```python
# access 5

nested[1][2][1]
```

# 2. Matrix-Style Access

```python
matrix = [[10, 20, 30], [40, 50, 60], [70, 80, 90]]
```

```python
# access the middle row

matrix[1]
```

```python
# access number 90

matrix[2][2]
```

```python
# access number 40

matrix[1][0]
```

# List Methods

# 1. using append()

```python
fruits = ['apple', 'banana']
```

```python
fruits.append(['kiwi', 'mango'])        # adds the list as one single item.
```

```python
fruits
```

# 2. using extend()

```python
fruits = ['apple', 'banana']
```

```python
fruits.extend(['kiwi', 'banana'])     # adds each item individually.
```

```python
fruits
```

# 3. using insert()

```python
nums = [10, 20, 40, 50]
```

```python
# insert 30 in correct numerical position.

nums.insert(2, 30)
print(nums)
```
