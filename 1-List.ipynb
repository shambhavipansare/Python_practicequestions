# 1. Deep Indexing

```python
data = ["a", "b", ["c", ["d", "e", "target"], "f"], "g"]
```

```python
data[2][1][2]      # extract the word target
```

# 2. Append & Extend

```python
colors = ["red", "blue"]
```

```python
colors.append(["green", "yellow"])      # add a list
```

```python
colors
```

```python
colors = ["red", "blue"]
```

```python
colors.extend(["green", "yellow"])      # add as elements
```

```python
colors
```
